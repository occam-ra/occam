# Minimal OCCAM Test Notebook
Simple cells to test functionality step by step

```python
# Cell 1: Imports and config
import pyoccam2 as pyoccam
import time

DATA_FILE = "dementia05.txt"
TARGET_STATE = "0"
print("Config loaded")
```

```python
# Cell 2: Initialize
manager = pyoccam.VBMManager()
if manager.init_from_command_line(["occam", DATA_FILE]):
    print(f"✓ Loaded {DATA_FILE}")
    print(f"  Sample size: {manager.get_sample_size()}")
else:
    print("ERROR loading data")
```

```python
# Cell 3: Configure
manager.set_report_separator(pyoccam.HTMLFORMAT)
manager.set_ref_model("bottom")
manager.set_fit_classifier_target(TARGET_STATE)
print("✓ Configured")
```

```python
# Cell 4: Search
print("Searching...")
search_report = manager.generate_search_report("full-up", 3, 3, False)
print(f"✓ Search done, {len(search_report)} chars")

# Get best models
best_bic = manager.get_best_model_by_bic()
print(f"Best BIC: {best_bic}")
```

```python
# Cell 5: Test fit - SIMPLE
print("Testing simplest fit...")
try:
    report = manager.generate_fit_report(best_bic, "0")
    #fit_report = manager.generate_fit_report(best_bic, "0")
    print(f"✓ Simple fit works: {len(report) if report else 0} chars")
except Exception as e:
    print(f"✗ Error: {e}")
```

```python
# Cell 5: Test fit - SIMPLE
print("Testing simplest fit...")
try:
    report = manager.generate_fit_report(best_bic, "0")
    fit_report = manager.generate_fit_report(best_bic, "0")
    print(f"✓ Simple fit works: {len(report) if report else 0} chars")
except Exception as e:
    print(f"✗ Error: {e}")
```

```python
# Cell 6: Test fit - WITH CONFUSION MATRIX
print("Testing fit with confusion matrix...")
try:
    report = manager.generate_fit_report("IV:Z", TARGET_STATE)
    if report and "Confusion" in report:
        print(f"✓ Confusion matrix works!")
    else:
        print(f"✓ Fit works but no confusion matrix")
except Exception as e:
    print(f"✗ Error: {e}")
```

```python
# Cell 7: Test fit - BEST MODEL
if best_bic:
    print(f"Testing fit for best model: {best_bic}")
    try:
        report = manager.generate_fit_report(best_bic, TARGET_STATE)
        if report:
            print(f"✓ SUCCESS! {len(report)} chars")
            with open("fit_test.txt", "w") as f:
                f.write(report)
            print("Saved to fit_test.txt")
        else:
            print("Empty report")
    except Exception as e:
        print(f"✗ Error: {e}")
else:
    print("No best model found")
```
